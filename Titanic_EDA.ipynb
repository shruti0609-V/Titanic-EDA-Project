# Titanic Dataset - Exploratory Data Analysis (EDA)

In this notebook, we perform basic EDA on the Titanic dataset using Pandas, Matplotlib, Seaborn, and Plotly.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
sns.set(style='whitegrid')
```

```python
# Load the dataset (make sure titanic.csv is in the same directory)
df = pd.read_csv('titanic.csv')
df.head()
```

```python
# Summary statistics and missing values
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
# Clean data
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
df.drop('Cabin', axis=1, inplace=True)
```

```python
# Histograms
df.hist(bins=20, figsize=(14, 8))
plt.suptitle("Histograms of Numerical Features")
plt.show()
```

```python
# Boxplot
plt.figure(figsize=(10,6))
sns.boxplot(x='Pclass', y='Age', data=df)
plt.title('Age Distribution Across Passenger Class')
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
# Pairplot
sns.pairplot(df[['Survived', 'Pclass', 'Age', 'Fare']], hue='Survived')
plt.show()
```

```python
# Countplots
sns.countplot(x='Survived', data=df)
plt.title('Survival Count')
plt.show()

sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Survival by Gender')
plt.show()
```

```python
# Skewness
print("Skewness of Age: ", df['Age'].skew())
print("Skewness of Fare: ", df['Fare'].skew())
```
